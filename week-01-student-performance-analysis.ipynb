```python
# Student Performance Analysis
#This notebook explores the relationship between students' study hours,attendance, and academic performance.
```

```python
import pandas as pd
```

```python
data = {
    "Name": ["Ada", "David", "Sarah", "John", "Michael"],
    "Study_Hours": [5, 2, 7, 3, 6],
    "Attendance": [90, 60, 95, 70, 88],
    "Programming": [85, 55, 92, 65, 80],
    "Math": [78, 61, 88, 68, 82]
}

df = pd.DataFrame(data)

df
```

```python
print("Number of rows and columns:", df.shape)
df.info()
```

```python
df.describe()
```

```python
print(df.loc[df["Study_Hours"].idxmax()])
```

```python
print(df.loc[df["Programming"].idxmax()])
```

```python
print(df["Programming"].mean())
```

```python
print(df["Attendance"].mean())
```

```python
print(df[["Name", "Study_Hours", "Programming"]].sort_values("Study_Hours"))
```

```python
correlation = df["Study_Hours"].corr(df["Programming"])
print("Study Hours vs Programming correlation:", correlation)
```

```python
## Observation
#The dataset shows a positive relationship between study hours and Programming scores. Students who studied more generally had higher Programming scores.
#However, this small dataset does not prove that study hours directly cause higher scores.
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Study_Hours"], df["Programming"])
plt.xlabel("Study Hours")
plt.ylabel("Programming Score")
plt.title("Study Hours vs Programming Score")
plt.show()
```

```python
# Conclusion
#From this dataset:
# Students who studied more generally had higher Programming scores.
# Students with higher attendance also tended to have higher Programming scores.
# The dataset shows relationships between these variables, but it is too small to establish causation.
# More data would be needed to make stronger conclusions.
#Michael studied 6 hours but scored 80, while Ada studied 5 hours and scored 85, showing that study hours alone don't completely explain performance
```
